Mean >>> the sum of observations divided by the total number of observations. It is also defined as average which is the sum divided by count.

```python
import numpy as np 

arr = [3,5,6]

mean = np.mean(arr)

print("Mean : ",mean)
```

Mode >>> The most frequently occurring value in the dataset. It’s useful for categorical data and in cases where knowing the most common choice is crucial.

```python
import scipy.stats as stats

# sample Data
arr = [1, 2, 2, 3]

# Mode
mode = stats.mode(arr)
print("Mode = ", mode)
```

Median : The median is the middle value in a sorted dataset. If the number of values is odd, it's the center value, if even, it's the average of the two middle values. It's often better than the mean for skewed data.

```python
import numpy as np 

arr = [1,2,3,4,5,6]

median = np.median(arr)

print("Median :", median)
```

Measure of Variability
Knowing not just where the data centers but also how it spreads out is important. Measures of variability, also called measures of dispersion, help us spot the spread or distribution of observations in a dataset.

1. Range : Range is a difference between the largest and smallest data point in our data set. The bigger the range the more spread the data and vice versa.
Syntax :
Range = Largest data value - smallest data value 

```python
import numpy as np 

arr = [2,5,8,12,56]

#finding maximum 
maximum = max(arr)

#finding minimum 
minimum = min(arr)

#range
range = maximum - minimum

print("Maximum = {}, minimum = {}, range = {}".format(maximum,minimum,range))
```

2.Variance: is defined as an average squared deviation from the mean. It is calculated by finding the difference between every data point and the average which is also known as the mean, squaring them, adding all of them and then dividing by the number of data points present in our data set.

```python
import statistics as st

arr = [2,3,4,5,6]

print("Varience ", st.variance(arr))
```

3.Standard deviation: Standard deviation is widely used to measure the extent of variation or dispersion in data. It's especially important when assessing model performance (e.g., residuals) or comparing datasets with different means.

```python
import statistics
arr = [1, 2, 3, 4, 5]
print("Std = ", (statistics.stdev(arr)))
```
