Basic terms of Statistics are : Data , Sample , Variable , Parameter , Population , Statistic 
There is two types of Statistics :
1. Descriptive Statistics 
  > Measure of Central Tendency >> Mean, Mode , Median
  > Measure of Variability >> Range, Varience , Standard deviation , Interquartile range(IQR)
  > Graphical representation >> Histogram, Boxplot, Pie Chart

2. Infferential Statistics :
   > Probability Distribution
   > Hypothesis testing 
   > Regression Analysis
   > Confidence Intervals
